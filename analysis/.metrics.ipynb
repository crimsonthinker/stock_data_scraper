- Equity and Book Value: https://corporatefinanceinstitute.com/resources/valuation/what-is-equity-value.
- Market Value = Market Cap.
- Market Value Per Share (MVPS) = Market Value / Number of oustanding shares.
- Book Value Per Share (BVPS): https://www.investopedia.com/terms/b/bvps.asp.
    - BVPS > Market value Per Share -> Undervalued.
- Price To Book Ratio (P/B Ratio): https://www.investopedia.com/investing/using-price-to-book-ratio-evaluate-companies.
    - P/B Ratio = MVPS / BVPS = MV / BV.
    - Low P/B -> Undervalued, or the company's asset value is overstated.
    - Low P/B also mean the company is earning a very poor return on its assets (ROA).
    - High P/B -> The stock price is trading at a premium to the company's book value/its stock is trading at [ratio] its book value/Overvalued.
    - Useful in practice when applied to caapital-intensive businesses.
    - P/B < 1 -> Undervalued.
    - P/B < 0 -> Liabilities > assets.
        - But not always bad.
        - Need to look into the company's financial situation and outlook for the future to understand what is creating the negative P/B ratio.
- Debt-to-equity Ratio (D/E Ratio): https://www.investopedia.com/terms/d/debtequityratio.asp.
    - D/E = Total Liabilities / Total Shareholders' Equity.
    - Assets = Liabilities + Shareholder Equity.
    - How much debt a company has taken on relative to the value of its assets net of liabilities.
    - A high D/E ratio is often associated with high investment risk; it means that a company relies primarily on debt financing.
    - Changes in long-term debt and assets tend to affect D/E ratio the most because the numbers involved tend to be larger than for short-term debt and short-term assets.
    - D/E < 1: Relatively safe.
    - D/E > 2: Risky.
    - Companies in some industries, such as utilities, consumer staples, and banking, typically have relatively high D/E ratios.
    - D/E < 0: 
        - The company is not taking advantage of debt financing and its tax advantages.
        - It has negative shareholder equity. In other words, the company’s liabilities exceed its assets. In most cases, this would be considered a sign of high risk and an incentive to seek bankruptcy protection.
    - DE ratio from Yahoo is by percentage (e.g. 42 -> 42% -> 0.42)
- Trailing Earnings Per Share (EPS): https://www.investopedia.com/terms/t/trailingeps.asp
    - Trailing: A value calculated on a rolling basis (usually 12 months).
    - EPS = Company's profit / Outstanding shares.
    - High EPS -> More profitable.
- Enterprise value: https://www.investopedia.com/terms/e/enterprisevalue.asp.
    - A more comprehensive alternative to market capitalization.
    - EV = MC + Total Debt - C (C: Cas and cash equivalents; the liquid assets of a company, but may ot include market securities).
- Earning-before-interest, taxes, depreciation, and amortization (EBITDA): https://www.investopedia.com/terms/e/ebitda.asp.
    - Represent the cash profit generated by the company's operations.
    - EBITDA = Net Income + Taxes = Interest Expense + D&A (Depreciation and amortization).
- Enterprise-value-to-EBITDA Ratio (EV/EBITDA): https://www.investopedia.com/ask/answers/072715/what-considered-healthy-evebitda.asp.
    - EV/EBITDA < 10: Healthy and above average.
    - High EV/EBITDA: The company is overvalued.
    - This metric can only apply it comparatively in a specific sector/industry.
- Enterprise-to-revenue (EV/R): https://www.investopedia.com/terms/e/ev-revenue-multiple.asp.
    - The lower the better.
    - EV/R has is that it can be used for companies that are yet to generate income or profits.
