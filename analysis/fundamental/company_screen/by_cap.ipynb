```python
mid_cap_query = f"""
    SELECT
        company_name,
        fundamental->>'industry' AS industry,
        stock_code,
        fundamental->>'marketCap' AS market_cap,
        fundamental->>'priceToBook' AS pb_ratio,
        fundamental->>'profitMargins' AS profit_margins,
        fundamental->>'trailingEps' AS earning_per_share,
        fundamental->>'enterpriseToEbitda' AS evebitda,
        fundamental->>'trailingPegRatio' AS peg_ratio,
        fundamental->>'debtToEquity' AS de_ratio
    FROM stock_info
    WHERE
        CAST(fundamental->>'marketCap' AS FLOAT) >= 100000000
        AND
        CAST(fundamental->>'marketCap' AS FLOAT) < 4000000000
        AND
        CAST(fundamental->>'priceToBook' AS FLOAT) >= 0.1
        AND
        CAST(fundamental->>'priceToBook' AS FLOAT) < 1
        AND
        CAST(fundamental->>'debtToEquity' AS FLOAT) < 15
"""
df = pd.read_sql_query(mid_cap_query, engine)
df
```
