# Australia

```python
engine = SQLModule.get_engine(country = 'australia')
query = f"""
    SELECT
        fundamental->>'sector' AS sector,
        fundamental->>'industry' AS industry
    FROM stock_info
"""
df = pd.read_sql_query(query, engine)
df[['sector', 'industry']].drop_duplicates().sort_values(by = 'sector')
```

```python
# Get comany section info of a company
engine = SQLModule.get_engine(country = 'vietnam')
stock_codes = ['ACB']
mid_cap_query = f"""
    SELECT
        stock_code,
        company_name,
        fundamental->>'sector' AS sector,
        fundamental->>'industry' AS industry
    FROM stock_info
    WHERE
        stock_code IN {tuple(stock_codes) if len(stock_codes) > 1 else f"('{stock_codes[0]}')"}
"""
comp = pd.read_sql_query(mid_cap_query, engine)
comp
```
