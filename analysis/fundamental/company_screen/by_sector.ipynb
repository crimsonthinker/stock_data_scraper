```python
from utils.generic_utils import SQLModule
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
engine = SQLModule.get_engine(country = 'australia')
```

```python
# Collect from 'company_section.ipynb'
SECTOR = 'Consumer Defensive'
```

```python
# Get distribution of metrics of company within the industry
query = f"""
    SELECT
        company_name,
        stock_code,
        fundamental->>'industry' AS industry,
        fundamental->>'marketCap' AS market_cap,
        fundamental->>'priceToBook' AS pb_ratio,
        fundamental->>'profitMargins' AS profit_margins,
        fundamental->>'trailingEps' AS earning_per_share,
        fundamental->>'enterpriseToEbitda' AS evebitda,
        fundamental->>'trailingPegRatio' AS peg_ratio,
        fundamental->>'debtToEquity' AS de_ratio
    FROM stock_info
    WHERE
        fundamental->>'sector' = '{SECTOR}'
"""
df = pd.read_sql_query(query, engine)
df
```

```python
# Distribution of market cap
fig = go.Figure()
fig.add_trace(
    go.Histogram(x = df['market_cap'].apply(lambda x : int(x)))
)
fig.update_yaxes(showgrid = True, gridcolor = 'gray', title = 'Number of company')
fig.update_xaxes(title = f'Market capital of company in sector "{SECTOR}"')
fig.update_layout(
    width = 1000,
    height = 500,
    plot_bgcolor = 'white',
    paper_bgcolor = 'white',
    font = dict(family = 'Lato', size = 20)
)
fig.show()
```

```python
# Screen companies by metrics
query = f"""
    SELECT
        company_name,
        stock_code,
        fundamental->>'industry' AS industry,
        
        fundamental->>'marketCap' AS market_cap,
        fundamental->>'priceToBook' AS pb_ratio,
        fundamental->>'profitMargins' AS profit_margins,
        fundamental->>'trailingEps' AS earning_per_share,
        fundamental->>'enterpriseToEbitda' AS evebitda,
        fundamental->>'trailingPegRatio' AS peg_ratio,
        fundamental->>'debtToEquity' AS de_ratio
    FROM stock_info
    WHERE
        fundamental->>'sector' = '{SECTOR}'
        AND
        CAST(fundamental->>'trailingEps' AS FLOAT) >= 200
        
"""
screen = pd.read_sql_query(query, engine).set_index('stock_code')
screen
```

```python
the_company = screen.loc['TNE']
the_company
```

```python
# Rerun the histogram, but add the company in it
fig = make_subplots(rows = 4, cols = 2)
metrics = ['market_cap', 'pb_ratio', 'profit_margins', 'earning_per_share', 'evebitda', 'peg_ratio', 'de_ratio']

for i,metric in enumerate(metrics):
    fig.add_trace(
        go.Histogram(
            x = df[metric].apply(lambda x : float(x) if not pd.isna(x) else None), 
            marker = dict(color = 'blue'),
            showlegend = False
        ),
        row = (i // 2) + 1, col = (i % 2) + 1
    )
    fig.update_xaxes(title = metric, row = (i // 2) + 1, col = (i % 2) + 1)
fig.update_yaxes(showgrid = True, gridcolor = 'gray', title = 'Number of company')

fig.update_layout(
    width = 1000,
    height = 1000,
    plot_bgcolor = 'white',
    paper_bgcolor = 'white',
    font = dict(family = 'Lato', size = 18)
)
fig_full = fig.full_figure_for_development()
for i,metric in enumerate(metrics):
    y_range = fig_full['layout'][f'yaxis{i + 1 if i > 0 else ""}']['range']
    if the_company[metric] is not None:
        fig.add_shape(
            type="line",
            x0 = the_company[metric], x1 = the_company[metric],  # X-position for the vertical line
            y0 = 0, y1 = y_range[1],      # Scaled Y-position (relative to subplot)
            line=dict(color="red", width=2),
            row = (i // 2) + 1, col = (i % 2) + 1
        )
fig.show()
```
